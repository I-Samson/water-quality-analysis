```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
print("Libraries loaded ✅")
```

```python
df = pd.read_csv("water_potability.csv")
print("Shape:", df.shape)
df.head()
```

```python
print("=== Dataset Info ===")
print(df.info())

print("\n=== Missing Values ===")
print(df.isnull().sum())
```

```python
df.fillna(df.mean(), inplace=True)
print("Missing values after fix:", df.isnull().sum().sum())
```

```python
df.describe().round(2)
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Potability', data=df, palette='Set2')
plt.title('Safe vs Unsafe Water Samples')
plt.xticks([0,1], ['Not Safe', 'Safe'])
plt.xlabel('Potability')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('potability_count.png')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.histplot(df['ph'], bins=30, kde=True, color='steelblue')
plt.title('Distribution of pH Levels in Water Samples')
plt.xlabel('pH')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('ph_distribution.png')
plt.show()
```

```python
plt.figure(figsize=(10,7))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap='coolwarm')
plt.title('Correlation Between Water Quality Parameters')
plt.tight_layout()
plt.savefig('correlation_heatmap.png')
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14,5))

for ax, col in zip(axes, ['ph', 'Hardness', 'Chloramines']):
    sns.boxplot(x='Potability', y=col, data=df, palette='Set3', ax=ax)
    ax.set_xticklabels(['Not Safe', 'Safe'])
    ax.set_title(f'{col} by Potability')

plt.suptitle('Key Parameters vs Water Safety', fontsize=14)
plt.tight_layout()
plt.savefig('boxplots.png')
plt.show()
```

```python
safe = df[df['Potability']==1].shape[0]
unsafe = df[df['Potability']==0].shape[0]
total = df.shape[0]

print(f"Total Samples: {total}")
print(f"Safe Water:    {safe} ({round(safe/total*100,1)}%)")
print(f"Unsafe Water:  {unsafe} ({round(unsafe/total*100,1)}%)")
print("\nKey Findings:")
print("- WHO safe pH range is 6.5–8.5. Our data shows most samples fall within this range.")
print("- Hardness and Chloramines show slight variation between safe and unsafe samples.")
print("- No single feature strongly predicts potability — a ML model would help further.")
```
